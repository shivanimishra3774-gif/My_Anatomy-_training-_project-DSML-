```python

# Range = Maximum - Minimum
Salary_range=df["Salary"].max()-df["Salary"].min()
print("Salary_range:",Salary_range)
```

```python

# Variance ask: how much do the values vary around mean ?
# Company A = 30k,31k,32k,33k,34k
# company B = 10k , 20k, 32k, 44k
# variance = (data point - mean)^2/
Salary_variance=df["Salary"].var()
print("Salary_variance:",Salary_variance)
```

```python

# Standard Deviation
Salary_std=df["Salary"].std()
print("Salary_std:",Salary_std)
```

```python
#Percentiles
# Instead of asking only for the average what is I want is I want to know where a

print("50th percentile",
      df["Salary"].quantile(0.50))
```

```python

q1 = df["Salary"].quantile(0.25)
q2 = df["Salary"].quantile(0.50)
q3 = df["Salary"].quantile(0.75)

print("Q1",q1)
print("Q2",q2)
print("Q3",q3)
```

```python

iqr = q3 - q1
print("IQR",iqr)
```

```python

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print("Lower Bound",lower_bound)
print("Upper Bound",upper_bound)
```

```python
# Find the outliners
outliners = df[(df["Salary"] < lower_bound) | (df["Salary"] > upper_bound)]
print(outliners)
```

```python

import random

total_rolls = int(input("Enter the total number of rolls: "))
count_4 = 0

for i in range(total_rolls):
    dice = random.randint(1, 6)

    if dice == 4:
        count_4 += 1

experimental_probability = count_4 / total_rolls

print("Total rolls:", total_rolls)
print("4 came:", count_4, "times")
print("Experimental probability:", experimental_probability)
print("Percentage:", experimental_probability * 100, "%")
```

```python

import pandas as pd
data = {
    "Customer" : range(1,16) ,
    "Visited_before" :["yes","no","yes","no","yes","no","no","no","yes","no","yes","no","yes","yes","yes"] ,
    "Added_to_card" : ["yes","no","no","yes","no","no","yes","no","yes","no","yes","no","yes","no","yes"] ,
    "Purchased" : ["yes","no","yes","no","yes","no","yes","no","no","no","yes","no","yes","no","yes"]

}
df = pd.DataFrame(data)
print(df)
```

```python

purchased_probability = (
    df["Purchased"] == "yes"
).mean()

print(purchased_probability)
```

```python

# what is the probability of purchase given that he cutomer added something to cart?
purchased_given_added_to_card = (
    df[df["Added_to_card"] == "yes"]["Purchased"] == "yes"
).mean()

#print(purchased_given_added_to_card)
print(purchased_given_added_to_card * 100, "%")
```
